**Round 1 — Basic DataFrame**


```python
import pandas as pd

data = {
    "name": ["Ali", "Ahmed", "Sara", "Hamza", "Ayesha", "Usman", "Zain"],
    "age": [18, 20, 19, 21, 18, 22, 20],
    "city": ["Lahore", "Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Multan"],
    "marks": [85, 72, 91, 65, 88, 76, 95],
    "passed": [True, True, True, False, True, True, True]
}

df = pd.DataFrame(data)
```

```python
df.head(5) #First Five rows
```

```python
df['name'] #Only name Column
```

```python
df[['name','marks']] #Display Name and marks column
```

```python
df.set_index('name', inplace=True) #Set name as index
```

```python
df
```

```python
df.reset_index(inplace = True) #Reset Index
```

```python
df
```

***Round_2:   Filtering***

```python
filt = df['marks']>80 #Find all students marks > 80
df.loc[filt]
```

```python
filt = df['age']==18 #Find students have age 18
df.loc[filt]
```

```python
df.loc[df['city']=='Lahore'] #Students from lahore
```

```python
filt = (df['city']=='Lahore') & (df['marks'] > 85) #From Lahore and marks more then 85
df.loc[filt]
```

```python
df.loc[(df["marks"]<70) | (df["marks"]>90)] #Marks less then 70 or greater then 90
```

***Round 3 — Updating***

```python
df
```

```python
df.loc[df['name'] == 'Ali','marks'] = 90 #Ali's Marks change 85 to 90
```

```python
df.loc[df['name'] == 'Hamza','passed'] = True #Hamza's passed value True from False
```

```python
df['marks'] = df['marks'].apply(lambda x: x+5) #Increase marks by 5
```

```python
df
```

```python
#Changing Zain's city from Multan to Lahore
df.loc[df['name']=='Zain','city'] = 'Lahore'
df
```

```python
#Q15
filt = (df['city']=='Lahore') & (df['age'] >= 18) & (df['marks'] >= 85)
df.loc[filt]
```
