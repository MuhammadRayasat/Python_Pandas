```python
import pandas as pd

data = {
    "name": ["Ali", "Ahmed", "Sara", "Hamza", "Ayesha",
             "Usman", "Zain", "Bilal", "Hira", "Daniyal"],
    "department": ["CS", "IT", "CS", "SE", "IT",
                   "CS", "SE", "IT", "CS", "SE"],
    "gender": ["M", "M", "F", "M", "F",
               "M", "M", "M", "F", "M"],
    "marks": [85, 72, 91, 65, 88, 76, 95, 72, 89, 81],
    "attendance": [90, 85, 95, 70, 92, 88, 98, 80, 94, 86],
    "city": ["Lahore", "Karachi", "Lahore", "Islamabad",
             "Lahore", "Karachi", "Multan", "Lahore",
             "Islamabad", "Karachi"]
}

df = pd.DataFrame(data)
```

```python
std_grp = df.groupby("department")
```

```python
result = std_grp.agg(
    avg_marks=('marks', 'mean'),
    max_marks=('marks', 'max'),
    avg_attendance=('attendance', 'mean')
)
display(result)
```

```python
std_grp.head()
```

```python
res = std_grp.agg(
    Total_Students = ('name','count'),
    Median_Marks = ('marks','median'),
    Min_Marks = ('marks','min')
)
display(res)
```
