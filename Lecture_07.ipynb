```python
import pandas as pd

data = {
    "name": ["Ali", "Ahmed", "Sara", "Hamza", "Ayesha", "Usman", "Zain", "Bilal"],
    "age": [18, 20, 19, 21, 18, 22, 20, 19],
    "city": ["Lahore", "Karachi", "Lahore", "Islamabad",
             "Lahore", "Karachi", "Multan", "Lahore"],
    "marks": [85, 72, 91, 65, 88, 76, 95, 72],
    "attendance": [90, 85, 95, 70, 92, 88, 98, 80]
}

df = pd.DataFrame(data)
```

```python
df=df.sort_values(by="marks")
```
