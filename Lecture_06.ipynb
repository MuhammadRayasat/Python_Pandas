```python
import pandas as pd

data = {
    "first_name": ["Ali", "Ahmed", "Sara", "Hamza", "Ayesha", "Usman", "Zain"],
    "last_name": ["Khan", "Malik", "Raza", "Ali", "Sheikh", "Butt", "Ahmed"],
    "age": [18, 20, 19, 21, 18, 22, 20],
    "city": ["Lahore", "Karachi", "Lahore", "Islamabad", "Lahore", "Karachi", "Multan"],
    "marks": [85, 72, 91, 65, 88, 76, 95]
}

df = pd.DataFrame(data)
```

```python
df["full_name"] = df["first_name"] + " " + df["last_name"] #Create a new column
```

```python
df
```

```python
df['result'] = df['marks'].apply(lambda x: 'Pass' if x >= 50 else 'Fail') #Q2
```

```python
df['age_group'] = df['age'].apply(lambda x: 'Teen' if x >= 20 else 'Adult')
```

```python
df[['first','last']]=df['full_name'].str.split(" ", expand=True) #Q3
```
